```python
#need to do some data cleaning because some of these columns are not of the correct data type
def clean_numbers(num):
    if num == "unavailable":
        num = np.nan
    elif num == "nr": 
        num = np.nan
    elif isinstance(num, str):
        num = num.replace(",", "")
        num = num.strip("<")
        num = num.strip(">")
        num = float(num)
    
    return num
```

```python
non_num_cols = data.dtypes[data.dtypes == 'object'].index.drop("U.S. State")
```

```python
for col in non_num_cols: 
    data[col] = data[col].apply(clean_numbers)

data.dtypes
```

```python
def bin_abortion_rate(rate): 
    if rate < 0: 
        return "-2-0%"
    elif rate < 2:
        return "0-2%"
    elif rate <= 4: 
        return "2-4%"
    elif rate <= 6: 
        return "4-6%"
    elif rate <= 8:
        return "6-8%"
    elif rate <= 10: 
        return "8-10%"
    elif rate <= 12: 
        return "10-12%"
    elif rate <= 14:
        return "12-14%"
    else: 
        return ">14%"
```

```python
add_code["bin_ab_R"] = add_code["Abortion rate (the no. of abortions per 1,000 women aged 15-17), by state of residence, 2017"].apply(bin_abortion_rate)
```

```python
#writing a new aggregation function
def pnp(grade):
    if grade in ["A", "B", "C"]:
        return "Pass"
    else:
        return "Fail"
```

```python


sex_ed_added = add_code.copy()
sex_ed_added = sex_ed_added.set_index("U.S. State")
sex_ed_added["Sex Ed Score"] = pd.Series(sex_ed_grades)
sex_ed_added.head()
```

```python
sex_ed_added["pass"] = sex_ed_added["Sex Ed Score"].apply(pnp)
sex_ed_added.columns
```

```python
#grouping by the passes and the non passes
mean_df = sex_ed_added.drop(columns = ['abbr', 'bin_ab_R', 'y', 'Sex Ed Score','Sex Ed Score']).groupby("pass").mean()

#change in abortion rate decreased in passing states and increased in failing states
#mwan aboirtion rate in 2017 is greater in passing states than failing states 
mean_df[['Abortion rate (the no. of abortions per 1,000 women aged 15-17), by state of residence, 2017',
       'Abortion rate (the no. of abortions per 1,000 women aged 15-19), by state of residence, 2017',
       'Abortion rate (the no. of abortions per 1,000 women aged 18-19), by state of residence, 2017',
       'Change in the no. of abortion clinics, 2017-2020',
       'Change in the no. of abortion providers, 2014-2017',
       'No. of abortion clinics, 2020', 'No. of abortion providers, 2017',
       'No. of abortions among women aged 15-17, by state of residence, 2017',
       'No. of abortions among women aged 15-19, by state of residence, 2017',
       'No. of abortions among women aged 18-19, by state of residence, 2017',
         'No. of abortions among women younger than 15, by state of residence, 2017']]
```

```python
sum_df = sex_ed_added.drop(columns = ['abbr', 'bin_ab_R', 'y', 'Sex Ed Score']).groupby("pass").sum()

sum_df[['Abortion rate (the no. of abortions per 1,000 women aged 15-17), by state of residence, 2017',
       'Abortion rate (the no. of abortions per 1,000 women aged 15-19), by state of residence, 2017',
       'Abortion rate (the no. of abortions per 1,000 women aged 18-19), by state of residence, 2017',
       'Change in the no. of abortion clinics, 2017-2020',
       'Change in the no. of abortion providers, 2014-2017',
       'No. of abortion clinics, 2020', 'No. of abortion providers, 2017',
       'No. of abortions among women aged 15-17, by state of residence, 2017',
       'No. of abortions among women aged 15-19, by state of residence, 2017',
       'No. of abortions among women aged 18-19, by state of residence, 2017',
       'No. of abortions among women younger than 15, by state of residence, 2017']]
```

```python
fig_map = {"Pass":"red", "Fail": "blue" }

fig = px.bar(mean_df,
    x = mean_df.index,
    y="Abortion rate (the no. of abortions per 1,000 women aged 15-19), by state of residence, 2017",
    range_y=[4.5, mean_df["Abortion rate (the no. of abortions per 1,000 women aged 15-19), by state of residence, 2017"].max()], 
    color = mean_df.index, color_discrete_map=fig_map,
    title =  "More Sex Ed, More Teen Abortions?", subtitle= "States with Stronger Sex Ed Programs Report Higher Abortion Rates Among Teens"
    )

fig.update_layout(xaxis_title = "Sex Education Score",
                   yaxis_title = "Mean Abortion Rate<br>(Number of Abortions per 1,000 women)<br>in 15-19 year olds, 2017",
                   showlegend = False,
                   plot_bgcolor="white")
fig.show()
```
